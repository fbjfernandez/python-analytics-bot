```python
!pip install wikipedia-api pandas matplotlib seaborn

import wikipediaapi
import pandas as pd
import re
import sqlite3
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files


wiki = wikipediaapi.Wikipedia('Latam_Analytics_Pipeline/1.0 (tu_correo@gmail.com)', 'es')

def extraer_texto_wikipedia(entidad):
    page = wiki.page(entidad)
    return page.text if page.exists() else ""


def procesar_datos_financieros(entidad, texto_crudo):
    oraciones = texto_crudo.split('. ')
    datos_estructurados = []

    patron_cifra = r'(\$?\d+(?:[.,]\d+)?\s*(?:millones|mil|billones|%|dólares|pasajeros|aviones)?)'

    categorias = {
        "Finanzas_y_Costos": ["dólares", "ingresos", "inversión", "pérdida", "us$", "millones", "compra"],
        "Operaciones_y_Flota": ["pasajeros", "flota", "aeronaves", "destinos", "vuelos", "rutas"]
    }

    for oracion in oraciones:
        oracion_limpia = oracion.lower().replace('\n', ' ').strip()

        for categoria, keywords in categorias.items():
            if any(kw in oracion_limpia for kw in keywords):
                cifras = re.findall(patron_cifra, oracion_limpia)
                if cifras:
                    datos_estructurados.append({
                        "Aerolinea": entidad,
                        "Categoria_Negocio": categoria,
                        "Metrica_Extraida": cifras[0],
                        "Contexto_Limpiado": oracion.replace('\n', ' ').strip()[:150] + "..."
                    })
                    break

    return datos_estructurados

aerolineas_latam = ["LATAM Airlines", "Avianca", "Copa Airlines", "Aerolíneas Argentinas"]
datos_totales = []

print("1. Conectando a la API de Wikipedia y extrayendo datos no estructurados...")
for aero in aerolineas_latam:
    texto = extraer_texto_wikipedia(aero)
    if texto:
        print(f"   -> Procesando texto de: {aero} ({len(texto)} caracteres)")
        datos_totales.extend(procesar_datos_financieros(aero, texto))

df_wiki = pd.DataFrame(datos_totales)

print("\n2. Cargando datos estructurados a motor SQL en memoria...")
conn = sqlite3.connect(':memory:')
df_wiki.to_sql('datos_inteligencia_mercado', conn, index=False)

query_sql = """
SELECT
    Aerolinea,
    SUM(CASE WHEN Categoria_Negocio = 'Finanzas_y_Costos' THEN 1 ELSE 0 END) as Menciones_Financieras,
    SUM(CASE WHEN Categoria_Negocio = 'Operaciones_y_Flota' THEN 1 ELSE 0 END) as Menciones_Operativas,
    COUNT(*) as Total_Datos_Extraidos
FROM datos_inteligencia_mercado
GROUP BY Aerolinea
ORDER BY Total_Datos_Extraidos DESC;
"""
df_analisis_sql = pd.read_sql_query(query_sql, conn)

print("\n--- RESULTADOS DEL ANÁLISIS SQL ---")
display(df_analisis_sql)

df_analisis_sql.set_index('Aerolinea')[['Menciones_Financieras', 'Menciones_Operativas']].plot(
    kind='bar', stacked=True, figsize=(10, 6), color=['#1f77b4', '#ff7f0e']
)
plt.title('Volumen de Datos Estructurados Extraídos de Wikipedia (Por Categoría)', fontsize=14)
plt.ylabel('Cantidad de KPIs Identificados')
plt.xlabel('Aerolínea Competidora')
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("analisis_competitivo_wikipedia.png")

df_wiki.to_csv("dataset_estructurado_wikipedia.csv", index=False)


df_wiki.to_excel("dataset_estructurado_wikipedia.xlsx", index=False)
files.download("dataset_estructurado_wikipedia.xlsx")
files.download("dataset_estructurado_wikipedia.csv")
files.download("analisis_competitivo_wikipedia.png")
```
